```python
plt.figure(figsize=(14, 8))  
corr_matrix = df[numeric_cols].corr()

# Create a mask to hide upper triangle (optional for readability)
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

# Drawing the heatmap with better aesthetics
sns.heatmap(corr_matrix, 
            annot=True, 
            cmap="coolwarm", 
            fmt=".2f", 
            linewidths=0.5, 
            mask=mask,  # Hide upper triangle
            annot_kws={"size": 10})  

plt.title("Correlation Heatmap", fontsize=14)
plt.show()
```

#### Taxes are directly proportional to total sales... as there is The strong correlation (close to 1) between Total Sales Tax @ 18% and Total Amount Excluding Taxes....confirms that taxes are consistently applied as a percentage of sales.

#### The correlations between Quantity_Product_1 and Quantity_Product_2, etc., are moderate (0.2 - 0.5).
#### Reason: Customers might not be consistently purchasing multiple products together.

```python
# Reduce sample size to avoid memory issues
sample_size = min(200, len(df))  # Take a maximum of 200 samples

sample_df = df[numerical_cols].sample(n=sample_size, random_state=42)

# Generate pairplot with reduced data
sns.pairplot(sample_df, diag_kind="kde", plot_kws={'alpha': 0.5})
plt.suptitle("Pairplot Analysis: Feature Relationships (Sampled Data)", fontsize=16)
plt.show()
```

```python
from sklearn.cluster import KMeans
# Use "Customer_NAME" instead of "Customer_ID"
if "Customer_NAME" in df.columns and "Total Amount Including Taxes" in df.columns:
    # Group by customer and calculate total spending
    customer_spending = df.groupby("Customer_NAME")["Total Amount Including Taxes"].sum().reset_index()
    customer_spending.columns = ["Customer_NAME", "Total_Spending"]

    # Apply K-Means clustering (2 clusters: High & Low Spenders)
    kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
    customer_spending["Segment"] = kmeans.fit_predict(customer_spending[["Total_Spending"]])

    # Map clusters to meaningful labels
    cluster_map = {0: "Low Spender", 1: "High Spender"}
    customer_spending["Segment"] = customer_spending["Segment"].map(cluster_map)

    # Show the first few segmented customers
    customer_spending.head(10)
else:
    print("Required columns (Customer_NAME, Total Amount Including Taxes) are missing from the dataset.")
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=customer_spending, x="Total_Spending", hue="Segment", bins=30, kde=True, palette="viridis")
plt.title("Customer Segmentation: High vs. Low Spenders", fontsize=16)
plt.xlabel("Total Spending")
plt.ylabel("Count of Customers")
plt.show()
```
